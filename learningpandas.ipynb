Merge, Join, and Concatenate Dataframes

```python
df4 = pd.read_csv(r"C:\Users\lunas\Downloads\EHR SYSTEM PROJECT\Excel Files\LOTR.csv")
df4
```

```python
df5=pd.read_csv(r"C:\Users\lunas\Downloads\EHR SYSTEM PROJECT\Excel Files\LOTR 2.csv")
df5
```

```python
df4.merge(df5)
```

```python
df4.merge(df5, how='inner', on=['FellowshipID', 'FirstName'])
```

```python
df4.merge(df5, how='outer', on=['FellowshipID', 'FirstName'])
```

```python
df4.merge(df5, how = 'left')
```

```python
df4.merge(df5, how = 'right')
```

```python
df4.merge(df5, how = 'cross')
```

```python
df4.join(df5, on= 'FellowshipID', how = 'outer', lsuffix='_left', rsuffix='_right')
```

```python
df6=df4.set_index('FellowshipID').join(df5.set_index('FellowshipID'), how='outer', lsuffix='_left', rsuffix='_right')
df6
```

```python
pd.concat([df4, df5], join='inner')  #joins what columns are similar in both dataframes and ignores the rest
```

```python
pd.concat([df4, df5], join='inner', axis = 1)  #joins dataframes side by side based on their indices
```
